# 1️⃣ What is Matplotlib?

**Matplotlib** is a Python library for creating charts and visualizations.

```python
import matplotlib.pyplot as plt
```

🧠 Think:

**Pandas → prepare data**

**Matplotlib → visualize data**


````python
import matplotlib.pyplot as plt
import numpy as np
````

# 2️⃣ Basic Line Plot

Line charts are useful for showing **trends over time**.


````python
days = [1, 2, 3, 4, 5]
sales = [100, 120, 150, 140, 180]

plt.plot(days, sales)

plt.show()
````

# 3️⃣ Titles, Labels & Grid

````python
days = [1, 2, 3, 4, 5]
sales = [100, 120, 150, 140, 180]

plt.plot(days, sales)

plt.title("Daily Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
````

# 4️⃣ Figure Size

`figsize=(width, height)` controls the size.


````python
plt.figure(figsize=(8, 4))

plt.plot(days, sales)

plt.title("Daily Sales")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
````

# 5️⃣ Scatter Plot

Scatter plots show the relationship between two numerical variables.


````python
height = [150, 160, 165, 170, 175, 180]
weight = [50, 58, 62, 68, 72, 80]

plt.scatter(height, weight)

plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")

plt.show()
````

# 6️⃣ Bar Chart

Use bar charts to compare categories.


````python
cities = ["Pune", "Mumbai", "Nashik", "Nagpur"]
sales = [500, 800, 350, 600]

plt.bar(cities, sales)

plt.title("Sales by City")
plt.xlabel("City")
plt.ylabel("Sales")

plt.show()
````

# 7️⃣ Horizontal Bar Chart

````python
plt.barh(cities, sales)

plt.title("Sales by City")
plt.xlabel("Sales")
plt.ylabel("City")

plt.show()
````

# 8️⃣ Histogram

A histogram shows the **distribution** of numerical data.

Example: distribution of student marks.


````python
marks = [45, 55, 60, 62, 65, 68, 70, 72, 75, 78,
         80, 82, 85, 88, 90, 92, 95, 96, 70, 74]

plt.hist(marks, bins=5)

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
````

# 9️⃣ Pie Chart

Useful for showing proportions of a whole.

⚠️ For many categories, a bar chart is usually easier to read.


````python
labels = ["IT", "Finance", "HR", "Sales"]
employees = [40, 25, 15, 20]

plt.pie(employees, labels=labels, autopct="%1.1f%%")

plt.title("Employees by Department")

plt.show()
````

# 🔟 Multiple Lines

````python
months = ["Jan", "Feb", "Mar", "Apr", "May"]

product_a = [100, 120, 150, 130, 170]
product_b = [90, 110, 140, 160, 180]

plt.plot(months, product_a, label="Product A")
plt.plot(months, product_b, label="Product B")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()
plt.grid(True)

plt.show()
````

# 1️⃣1️⃣ Markers & Line Styles

````python
plt.plot(
    months,
    product_a,
    marker="o",
    linestyle="--",
    label="Product A"
)

plt.title("Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.show()
````

# 1️⃣2️⃣ Subplots

Subplots allow multiple charts in one figure.

Syntax:

```python
plt.subplots(rows, columns)
```


````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# First chart
axes[0].plot(months, product_a)
axes[0].set_title("Product A")

# Second chart
axes[1].bar(cities, sales)
axes[1].set_title("Sales by City")

plt.tight_layout()
plt.show()
````

# 1️⃣3️⃣ Saving a Chart

````python
plt.plot(months, product_a)
plt.title("Product A Sales")

# Save the chart to a file
plt.savefig("product_a_sales.png", dpi=300, bbox_inches="tight")

plt.show()
````

# 🎯 Practical Example — Monthly Sales

````python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12000, 15000, 13500, 18000, 21000, 19500]

plt.figure(figsize=(9, 5))

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")

plt.grid(True)
plt.tight_layout()

plt.show()
````
