# 2️⃣ Variables

A variable is simply a **name that stores a value**.

```python
name = "Shubham"
age = 22
```

Think:

`name → "Shubham"`

`age → 22`


````python
# Creating variables
name = "Shubham"
age = 22
height = 5.6

print(name)
print(age)
print(height)

# Python automatically understands the data type.
````

### Variable naming rules

✅ Good:
```python
student_name = "Rahul"
age = 20
marks1 = 85
```

❌ Bad:
```python
1name = "Rahul"       # cannot start with a number
student-name = "R"    # '-' is interpreted as minus
```

💡 Python variable names are **case-sensitive**:
`age` and `Age` are different variables.

# 5️⃣ Operators

### Arithmetic operators

`+` addition  
`-` subtraction  
`*` multiplication  
`/` division  
`//` floor division  
`%` remainder  
`**` power


````python
a = 10
b = 3

print(a + b)   # addition
print(a - b)   # subtraction
print(a * b)   # multiplication
print(a / b)   # normal division
print(a // b)  # floor division
print(a % b)   # remainder
print(a ** b)  # 10 raised to power 3
````

### Comparison operators

These return `True` or `False`.

`>` greater than  
`<` less than  
`>=` greater than/equal  
`<=` less than/equal  
`==` equal to  
`!=` not equal to


````python
age = 22

print(age > 18)
print(age == 22)
print(age != 30)
print(age <= 25)
````

### Logical operators

- `and` → both conditions must be True
- `or` → at least one condition must be True
- `not` → reverses True/False


````python
age = 22
has_id = True

print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
````

# 6️⃣ Strings

A string is text.

```python
name = "Shubham"
```

Strings support indexing:

```text
Python
012345
```

The first character is always at index `0`.


````python
text = "Python"

print(text[0])   # P
print(text[1])   # y
print(text[-1])  # n

# Slicing
print(text[0:3])  # Pyt
print(text[2:])   # thon
print(text[:4])   # Pyth
````

### Useful string methods


````python
text = "  hello python  "

print(text.upper())       # uppercase
print(text.lower())       # lowercase
print(text.strip())       # remove spaces at both ends
print(text.replace("python", "world"))

# split() converts a string into a list
sentence = "Python is easy"
print(sentence.split())
````

### f-strings ⭐

The easiest way to combine variables and text.


````python
name = "Shubham"
age = 22

message = f"My name is {name} and I am {age} years old."

print(message)
````

# 7️⃣ Lists

A list stores **multiple values**.

```python
numbers = [10, 20, 30]
```

Lists are:
- Ordered
- Mutable → can be changed
- Able to contain different data types


````python
numbers = [10, 20, 30, 40]

print(numbers)
print(numbers[0])
print(numbers[-1])

# Change an item
numbers[1] = 200

print(numbers)
````

### Common list methods


````python
fruits = ["apple", "banana", "mango"]

fruits.append("orange")       # add at the end
fruits.insert(1, "grapes")    # add at a specific position
fruits.remove("banana")       # remove a value

print(fruits)
print(len(fruits))            # number of items
````

# 8️⃣ Tuples

A tuple looks like a list, but it is **immutable**.

```python
coordinates = (10, 20)
```

Use tuples when the values should not be changed.


````python
coordinates = (10, 20)

print(coordinates[0])
print(coordinates[1])

# coordinates[0] = 100
# ❌ This would give an error because tuples cannot be changed.
````

# 9️⃣ Sets

A set stores **unique values**.

Duplicates are automatically removed.


````python
numbers = {1, 2, 2, 3, 3, 4}

print(numbers)

# Add an item
numbers.add(5)

print(numbers)
````

# 🔟 Dictionaries

A dictionary stores data as:

**key → value**

Example:

```python
student = {
    "name": "Rahul",
    "age": 21
}
```


````python
student = {
    "name": "Rahul",
    "age": 21,
    "marks": 85
}

print(student["name"])
print(student["marks"])

# Add a new key
student["city"] = "Pune"

print(student)
````

### Useful dictionary methods


````python
student = {"name": "Rahul", "age": 21}

print(student.keys())    # all keys
print(student.values())  # all values
print(student.items())   # key-value pairs

# get() is useful when a key may not exist
print(student.get("name"))
print(student.get("phone", "Not available"))
````

# 1️⃣1️⃣ `if / elif / else`

Use conditions when your program needs to **make decisions**.

```text
if condition:
    do something
else:
    do something else
```

⚠️ Python uses **indentation** to define blocks.


````python
marks = 72

if marks >= 90:
    print("Grade A+")
elif marks >= 75:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Needs improvement")
````

# 1️⃣2️⃣ `for` Loop

Use a `for` loop when you want to repeat something for each item.

`range()` is commonly used for numbers.


````python
# Print numbers from 0 to 4
for i in range(5):
    print(i)
````

### `range(start, stop, step)`

⚠️ The `stop` value is **not included**.


````python
for i in range(1, 11):
    print(i)

print("--- Even numbers ---")

for i in range(2, 11, 2):
    print(i)
````

### Loop through a list


````python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print("I like", fruit)
````

# 1️⃣3️⃣ `while` Loop

A `while` loop keeps running **while a condition is True**.

Be careful: if the condition never becomes False, you create an **infinite loop**.


````python
count = 1

while count <= 5:
    print(count)
    count += 1  # increase count so the loop eventually stops
````

# 1️⃣4️⃣ `break`, `continue`, `pass`

### `break`
Stops the loop completely.

### `continue`
Skips the current iteration.

### `pass`
Does nothing; useful as a placeholder.


````python
for i in range(1, 11):

    if i == 5:
        break

    print(i)
````

````python
for i in range(1, 6):

    if i == 3:
        continue

    print(i)
````

# 1️⃣5️⃣ Functions ⭐

A function is a reusable block of code.

```python
def function_name():
    # code
```

Why functions?

👉 Avoid repeating code  
👉 Make programs easier to understand  
👉 Make code reusable


````python
def greet():
    print("Hello!")

greet()
greet()
````

### Function parameters

Parameters allow us to send information into a function.


````python
def greet(name):
    print(f"Hello {name}!")

greet("Shubham")
greet("Rahul")
````

### `return`

`return` sends a result back from the function.


````python
def add(a, b):
    result = a + b
    return result

answer = add(10, 20)

print(answer)
````

### Default arguments


````python
def greet(name="Student"):
    print(f"Hello {name}!")

greet()
greet("Shubham")
````

# 1️⃣6️⃣ Scope

A variable created **inside** a function is usually local to that function.

A variable created outside is in the outer/global scope.


````python
name = "Global"

def test():
    name = "Local"
    print("Inside:", name)

test()

print("Outside:", name)
````

# 1️⃣7️⃣ List Comprehension

A compact way to create a list.

Normal approach:
```python
squares = []
for i in range(5):
    squares.append(i * i)
```

List comprehension:
```python
squares = [i * i for i in range(5)]
```


````python
# Normal loop
squares = []

for i in range(1, 6):
    squares.append(i ** 2)

print(squares)

# Same idea using list comprehension
squares2 = [i ** 2 for i in range(1, 6)]

print(squares2)
````

# 1️⃣9️⃣ Exception Handling

Programs can encounter errors.

Use `try / except` when you want to handle an error gracefully.


````python
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)

except ValueError:
    print("Please enter a valid integer.")
````

### Why is this useful?

Without exception handling, invalid input can stop your program.

With `try/except`, we can handle expected errors and keep the program running.


# 2️⃣0️⃣ File Handling

Python can read and write files.

### Writing a file


````python
# "w" = write mode
with open("example.txt", "w") as file:
    file.write("Hello Python!")

print("File created.")
````

### Reading a file

`with open(...)` automatically closes the file after use.


````python
with open("example.txt", "r") as file:
    content = file.read()

print(content)
````

# 2️⃣1️⃣ Useful Built-in Functions

Some functions you'll use constantly:


````python
numbers = [10, 20, 30, 40, 50]

print(len(numbers))  # number of items
print(sum(numbers))  # total
print(max(numbers))  # largest
print(min(numbers))  # smallest
print(sorted(numbers, reverse=True))  # sorted descending
````

### `enumerate()`

Useful when you need both the **index** and the **value**.


````python
fruits = ["apple", "banana", "mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
````

# 2️⃣2️⃣ A Small Real Example

Let's combine variables, input, conditions and functions.

### 🎯 Student Marks Calculator


````python
def calculate_result(marks):
    # Calculate the average of all subjects
    average = sum(marks) / len(marks)

    print(f"Average marks: {average:.2f}")

    # Decide the result based on average
    if average >= 75:
        print("Grade: A")
    elif average >= 60:
        print("Grade: B")
    elif average >= 40:
        print("Grade: C")
    else:
        print("Result: Fail")


marks = [80, 72, 91, 68, 85]

calculate_result(marks)
````
