# 1️⃣ What is NumPy?

**NumPy = Numerical Python**

It provides fast, efficient arrays and mathematical operations.

```python
import numpy as np
```

🧠 **Remember:** Pandas is built heavily on top of NumPy concepts.


````python
import numpy as np

# Create a simple NumPy array
numbers = np.array([10, 20, 30, 40, 50])

print(numbers)
print(type(numbers))
````

# 2️⃣ Creating Arrays

````python
import numpy as np

# 1-dimensional array
a = np.array([1, 2, 3, 4])

# 2-dimensional array
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(a)
print(matrix)

# Arrays filled with zeros
print(np.zeros(5))

# Arrays filled with ones
print(np.ones((2, 3)))

# Sequence of numbers
print(np.arange(0, 10, 2))
````

# 3️⃣ Array Properties

Important properties:

- `shape` → dimensions
- `ndim` → number of dimensions
- `size` → total elements
- `dtype` → data type


````python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Total elements:", arr.size)
print("Data type:", arr.dtype)
````

# 4️⃣ Indexing & Slicing

Like Python lists, NumPy starts indexing from `0`.

For a 2D array:

```text
row 0 → [10, 20, 30]
row 1 → [40, 50, 60]
```


````python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(arr[0, 1])     # row 0, column 1
print(arr[1, 2])     # row 1, column 2

print(arr[0, :])     # first row
print(arr[:, 1])     # second column

print(arr[0:2, 1:3]) # rows 0-1, columns 1-2
````

# 5️⃣ Reshaping

`reshape()` changes the shape without changing the values.

⚠️ The number of elements must remain the same.


````python
arr = np.arange(1, 13)

print(arr)
print("Original shape:", arr.shape)

matrix = arr.reshape(3, 4)

print(matrix)
print("New shape:", matrix.shape)
````

# 6️⃣ Arithmetic & Broadcasting

NumPy performs operations **element by element**.

This is one of its biggest advantages over normal Python lists.


````python
prices = np.array([100, 200, 300, 400])

print(prices + 10)
print(prices * 2)
print(prices / 10)

# Broadcasting:
# NumPy automatically applies 18% GST to every value.
gst_prices = prices * 1.18

print(gst_prices)
````

# 7️⃣ Aggregations

Common functions:

`sum()` · `mean()` · `median()` · `min()` · `max()` · `std()`


````python
marks = np.array([72, 85, 91, 64, 78])

print("Total:", np.sum(marks))
print("Average:", np.mean(marks))
print("Median:", np.median(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
print("Std deviation:", np.std(marks))
````

### Aggregation by axis

For a 2D array:

- `axis=0` → work down the rows / column-wise
- `axis=1` → work across columns / row-wise


````python
marks = np.array([
    [80, 70, 90],
    [60, 85, 75],
    [90, 95, 88]
])

print("Column averages:", np.mean(marks, axis=0))
print("Row averages:", np.mean(marks, axis=1))
````

# 8️⃣ Boolean Filtering

Create a condition and use it to select values.

This is extremely important for data analysis.


````python
marks = np.array([45, 78, 92, 33, 67, 88])

# Boolean condition
passed = marks >= 50

print(passed)

# Use the condition to filter the array
print(marks[marks >= 50])

# Multiple conditions
print(marks[(marks >= 50) & (marks < 90)])
````

# 9️⃣ Random Numbers

Useful for simulations, testing and ML experiments.


````python
# Generate random integers from 1 to 99
numbers = np.random.randint(1, 100, size=10)

print(numbers)

# Random decimal values between 0 and 1
print(np.random.rand(5))
````

# 🔟 Useful NumPy Functions

````python
arr = np.array([5, 2, 8, 1, 9, 3])

print("Sorted:", np.sort(arr))
print("Unique:", np.unique([1, 2, 2, 3, 3, 3]))
print("Absolute:", np.abs([-10, -5, 2]))
print("Square root:", np.sqrt([4, 9, 16]))
print("Power:", np.power([2, 3, 4], 2))
````

# 1️⃣1️⃣ `NaN` — Missing Numerical Values

`np.nan` represents a missing numerical value.

Be careful: normal equality checks do not work as expected with `NaN`.


````python
data = np.array([10, 20, np.nan, 40, 50])

print("Average ignoring NaN:", np.nanmean(data))
print("Sum ignoring NaN:", np.nansum(data))
````

# 🎯 Practical Example — Student Marks

````python
students = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [92, 88, 95],
    [45, 55, 48]
])

# Average marks for each student
averages = np.mean(students, axis=1)

print("Student averages:", averages)

# Students whose average is >= 60
print("Passed:", averages >= 60)

# Highest subject-wise marks
print("Highest in each subject:", np.max(students, axis=0))
````
