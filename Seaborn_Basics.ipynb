# 1️⃣ What is Seaborn?

**Seaborn** is a statistical visualization library built on top of Matplotlib.

It works especially well with **Pandas DataFrames**.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

🧠 Simple idea:

**Matplotlib → general-purpose plotting**

**Seaborn → statistical/data-analysis visualizations**


````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load one of Seaborn's example datasets
tips = sns.load_dataset("tips")

print(tips.head())
````

# 2️⃣ Inspect the Dataset

````python
print(tips.shape)
print(tips.columns)
print(tips.info())
print(tips.describe())
````

# 3️⃣ Scatter Plot

Scatter plots help study relationships between numerical variables.


````python
sns.scatterplot(data=tips, x="total_bill", y="tip")

plt.title("Total Bill vs Tip")
plt.show()
````

### Add a categorical variable with `hue`

This is one of Seaborn's useful features.


````python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex"
)

plt.title("Bill vs Tip by Gender")
plt.show()
````

# 4️⃣ Line Plot

````python
# Example dataset for a time-like trend
sales = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 140, 180, 210, 195]
})

sns.lineplot(data=sales, x="Month", y="Sales", marker="o")

plt.title("Monthly Sales")
plt.show()
````

# 5️⃣ Bar Plot

Useful for comparing an average/statistical value across categories.


````python
sns.barplot(data=tips, x="day", y="total_bill")

plt.title("Average Bill by Day")
plt.show()
````

# 6️⃣ Count Plot

A count plot shows **how many observations** belong to each category.


````python
sns.countplot(data=tips, x="day")

plt.title("Number of Orders by Day")
plt.show()
````

# 7️⃣ Histogram + KDE

A histogram shows distribution.

KDE = **Kernel Density Estimate**, a smooth estimate of the distribution.


````python
sns.histplot(
    data=tips,
    x="total_bill",
    kde=True
)

plt.title("Distribution of Total Bills")
plt.show()
````

# 8️⃣ Box Plot ⭐

Box plots are excellent for identifying:

- Median
- Quartiles
- Spread
- Potential outliers


````python
sns.boxplot(data=tips, x="day", y="total_bill")

plt.title("Bill Distribution by Day")
plt.show()
````

# 9️⃣ Violin Plot

A violin plot combines ideas from a box plot and a distribution plot.


````python
sns.violinplot(data=tips, x="day", y="total_bill")

plt.title("Bill Distribution by Day")
plt.show()
````

# 🔟 Heatmap

Heatmaps use intensity to show values in a matrix.

A common use in Data Analytics/ML is a **correlation matrix**.


````python
# Select numerical columns and calculate correlation
correlation = tips.select_dtypes("number").corr()

print(correlation)

sns.heatmap(
    correlation,
    annot=True
)

plt.title("Correlation Heatmap")
plt.show()
````

# 1️⃣1️⃣ Pair Plot

A pair plot shows relationships between multiple numerical variables at once.


````python
sns.pairplot(
    tips,
    hue="sex"
)

plt.show()
````

# 1️⃣2️⃣ Categorical Analysis

Seaborn makes it easy to compare categories with `hue`.


````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.title("Average Bill by Day and Gender")
plt.show()
````

# 1️⃣3️⃣ Styling

Seaborn provides simple themes.

Common choices include:

```python
sns.set_theme(style="whitegrid")
sns.set_theme(style="darkgrid")
```


````python
sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)

plt.title("Bill vs Tip")
plt.show()
````

# 1️⃣4️⃣ Pandas + Seaborn Workflow

This is a very common Data Analytics workflow:

```text
CSV / Excel
     ↓
Pandas
     ↓
Clean data
     ↓
Analyze
     ↓
Seaborn / Matplotlib
     ↓
Insights
```


````python
# Create an analysis-ready example
summary = (
    tips.groupby("day", observed=True)["total_bill"]
    .mean()
    .reset_index()
)

print(summary)

sns.barplot(
    data=summary,
    x="day",
    y="total_bill"
)

plt.title("Average Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Bill")

plt.show()
````

# 🎯 Mini EDA Questions

Using the `tips` dataset, try answering:

1. Which day has the highest average bill?
2. Is the average tip different between lunch and dinner?
3. Which day has the most observations?
4. Are larger bills generally associated with larger tips?
5. Are there potential outliers in total bills?
6. Does the relationship between bill and tip differ by smoker/non-smoker?


````python
# Start exploring!

print("Average bill by day:")
print(tips.groupby("day", observed=True)["total_bill"].mean())

print("\nAverage tip by time:")
print(tips.groupby("time", observed=True)["tip"].mean())

print("\nAverage bill by smoker:")
print(tips.groupby("smoker", observed=True)["total_bill"].mean())
````
