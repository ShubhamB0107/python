# 1️⃣ What is Pandas?

**Pandas** is the main Python library for working with structured/tabular data.

Think of a Pandas DataFrame like an **Excel table inside Python**.

```python
import pandas as pd
```


````python
import pandas as pd

print(pd.__version__)
````

# 2️⃣ Series

A Series is basically **one column of data**.


````python
marks = pd.Series([80, 75, 91, 68, 85])

print(marks)
print(marks.mean())
print(marks.max())
````

# 3️⃣ DataFrame

A DataFrame is a table with rows and columns.


````python
data = {
    "Name": ["Amit", "Rahul", "Priya", "Sneha"],
    "Age": [21, 22, 20, 23],
    "Marks": [82, 75, 91, 68]
}

df = pd.DataFrame(data)

print(df)
````

# 4️⃣ Reading CSV / Excel

For real data analysis, you'll frequently load files.

```python
pd.read_csv("file.csv")
pd.read_excel("file.xlsx")
```

⚠️ Excel support may require the appropriate engine/package.


````python
# Example:
# df = pd.read_csv("sales.csv")
# df = pd.read_excel("sales.xlsx")

# For this notebook, create a sample DataFrame instead.
data = {
    "Name": ["Amit", "Rahul", "Priya", "Sneha", "Amit"],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Pune"],
    "Age": [21, 22, 20, 23, 21],
    "Marks": [82, 75, 91, 68, 82],
    "Salary": [35000, 42000, 50000, 38000, 35000]
}

df = pd.DataFrame(data)
df
````

# 5️⃣ Inspecting Data

These are some of the **first commands** you should run after loading a dataset.


````python
print(df.head())       # first 5 rows
print(df.tail())       # last 5 rows
print(df.shape)        # rows, columns
print(df.columns)      # column names
print(df.dtypes)       # data types
print(df.info())       # detailed structure
````

### Statistics with `describe()`

````python
print(df.describe())
````

# 6️⃣ Selecting Columns

````python
print(df["Name"])

# Select multiple columns
print(df[["Name", "Marks", "Salary"]])
````

# 7️⃣ Selecting Rows — `loc` and `iloc`

- `loc` → label-based selection
- `iloc` → position-based selection


````python
# First row by position
print(df.iloc[0])

# First three rows
print(df.iloc[0:3])

# Specific row + columns
print(df.loc[0, "Name"])

# Multiple columns
print(df.loc[:, ["Name", "Marks"]])
````

# 8️⃣ Filtering Data

Filtering is one of the most important Pandas skills.

Think:

`df[condition]`


````python
# Students with marks >= 80
high_scorers = df[df["Marks"] >= 80]

print(high_scorers)

# Students from Pune
print(df[df["City"] == "Pune"])

# Multiple conditions
print(df[(df["Marks"] >= 80) & (df["Age"] <= 22)])
````

# 9️⃣ Sorting

````python
# Sort by marks
print(df.sort_values("Marks"))

# Highest marks first
print(df.sort_values("Marks", ascending=False))

# Sort by multiple columns
print(df.sort_values(["City", "Marks"], ascending=[True, False]))
````

# 🔟 Adding & Changing Columns

````python
# Create a new column
df["Passed"] = df["Marks"] >= 40

# Calculate bonus
df["Bonus"] = df["Salary"] * 0.10

print(df)
````

# 1️⃣1️⃣ Missing Values

Real datasets often contain missing data.

Important functions:

- `isna()`
- `notna()`
- `fillna()`
- `dropna()`


````python
data = {
    "Name": ["A", "B", "C", "D"],
    "Marks": [80, None, 90, 70],
    "City": ["Pune", "Mumbai", None, "Pune"]
}

missing_df = pd.DataFrame(data)

print(missing_df.isna())
print(missing_df.isna().sum())
````

````python
# Fill missing marks with the column average
missing_df["Marks"] = missing_df["Marks"].fillna(
    missing_df["Marks"].mean()
)

# Fill missing city with a label
missing_df["City"] = missing_df["City"].fillna("Unknown")

print(missing_df)
````

# 1️⃣2️⃣ Duplicates

````python
print(df.duplicated())

# Remove duplicate rows
clean_df = df.drop_duplicates()

print(clean_df)
````

# 1️⃣3️⃣ GroupBy ⭐

`groupby()` is extremely important for Data Analytics.

Example question:

> What is the average salary in each city?


````python
print(df.groupby("City")["Salary"].mean())
````

### Multiple aggregations

````python
summary = df.groupby("City").agg(
    average_salary=("Salary", "mean"),
    highest_salary=("Salary", "max"),
    employee_count=("Name", "count"),
    average_marks=("Marks", "mean")
)

print(summary)
````

# 1️⃣4️⃣ `value_counts()`

Useful for categorical columns.


````python
print(df["City"].value_counts())
print(df["Passed"].value_counts())
````

# 1️⃣5️⃣ `apply()`

Use `apply()` when you want to run a function on values.


````python
def grade(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 75:
        return "A"
    elif marks >= 60:
        return "B"
    else:
        return "C"

df["Grade"] = df["Marks"].apply(grade)

print(df[["Name", "Marks", "Grade"]])
````

# 1️⃣6️⃣ Combining DataFrames

### `concat()`
Stacks DataFrames vertically or horizontally.

### `merge()`
Similar to a database/SQL JOIN.


````python
employees = pd.DataFrame({
    "EmployeeID": [1, 2, 3],
    "Name": ["Amit", "Rahul", "Priya"]
})

departments = pd.DataFrame({
    "EmployeeID": [1, 2, 3],
    "Department": ["IT", "Finance", "HR"]
})

merged = pd.merge(employees, departments, on="EmployeeID")

print(merged)
````

# 1️⃣7️⃣ Dates

Convert text dates into Pandas datetime objects using `pd.to_datetime()`.


````python
dates = pd.DataFrame({
    "Date": ["2026-01-10", "2026-02-15", "2026-03-20"],
    "Sales": [10000, 15000, 12000]
})

dates["Date"] = pd.to_datetime(dates["Date"])

dates["Year"] = dates["Date"].dt.year
dates["Month"] = dates["Date"].dt.month
dates["Month_Name"] = dates["Date"].dt.month_name()

print(dates)
````
